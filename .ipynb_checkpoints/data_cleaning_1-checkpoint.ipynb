```python
restaurants.drop_duplicates(inplace=True)
```

```python
places["Weekly Off"] = places["Weekly Off"].fillna("No Weekly Off")
```

```python
places.columns
```

```python
restaurants.isnull().sum()
```

```python
restaurants.drop_duplicates(inplace=True)
```

```python
goibibo.isnull().sum()
```

```python
goibibo.drop(
    columns=["hotel_brand", "room_area", "qts"],
    inplace=True
)
```

```python
goibibo["additional_info"] = goibibo["additional_info"].fillna("Not Available")
goibibo["guest_recommendation"] = goibibo["guest_recommendation"].fillna("Not Rated")
goibibo["hotel_description"] = goibibo["hotel_description"].fillna("No Description")
goibibo["hotel_facilities"] = goibibo["hotel_facilities"].fillna("Unknown")
goibibo["room_facilities"] = goibibo["room_facilities"].fillna("Unknown")
goibibo["similar_hotel"] = goibibo["similar_hotel"].fillna("None")
goibibo["point_of_interest"] = goibibo["point_of_interest"].fillna("Unknown")
goibibo["area"] = goibibo["area"].fillna("Unknown")
goibibo["locality"] = goibibo["locality"].fillna("Unknown")
```

```python
goibibo["site_review_rating"] = goibibo["site_review_rating"].fillna(0)
goibibo["site_review_count"] = goibibo["site_review_count"].fillna(0)
goibibo["review_count_by_category"] = goibibo["review_count_by_category"].fillna("Not Available")
```

```python
goibibo.isnull().sum()
```

```python
# Fill categorical columns
cleartrip["hotel_description"] = cleartrip["hotel_description"].fillna("No Description")
cleartrip["hotel_facilities"] = cleartrip["hotel_facilities"].fillna("Unknown")
cleartrip["room_facilities"] = cleartrip["room_facilities"].fillna("Unknown")
cleartrip["similar_hotel"] = cleartrip["similar_hotel"].fillna("None")
cleartrip["property_type"] = cleartrip["property_type"].fillna("Hotel")

# Fill location columns
cleartrip["area"] = cleartrip["area"].fillna("Unknown")
cleartrip["locality"] = cleartrip["locality"].fillna("Unknown")
cleartrip["state"] = cleartrip["state"].fillna("Unknown")

# Fill numerical review columns
cleartrip["tad_review_count"] = cleartrip["tad_review_count"].fillna(0)
cleartrip["tad_review_rating"] = cleartrip["tad_review_rating"].fillna(0)
cleartrip["tad_stay_review_rating"] = cleartrip["tad_stay_review_rating"].fillna(0)
cleartrip["tripadvisor_seller_rating"] = cleartrip["tripadvisor_seller_rating"].fillna(0)
```

```python
cleartrip.isnull().sum()
```

```python
cleartrip["hotel_star_rating"] = cleartrip["hotel_star_rating"].fillna(
    cleartrip["hotel_star_rating"].median()
)
```

```python
# Fill text columns
cleartrip["address"] = cleartrip["address"].fillna("Unknown")
cleartrip["image_urls"] = cleartrip["image_urls"].fillna("No Image")

# Fill room type
cleartrip["room_type"] = cleartrip["room_type"].fillna(
    cleartrip["room_type"].mode()[0]
)

# Remove rows with missing coordinates
cleartrip.dropna(subset=["latitude", "longitude"], inplace=True)
```

```python
cleartrip.isnull().sum()
```

```python
places.to_csv("clean_places.csv", index=False)
restaurants.to_csv("clean_restaurants.csv", index=False)
goibibo.to_csv("clean_goibibo.csv", index=False)
cleartrip.to_csv("clean_cleartrip.csv", index=False)
```
